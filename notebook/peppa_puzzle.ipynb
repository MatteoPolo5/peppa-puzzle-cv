```python
# Convert to grayscale
gray = cv.cvtColor(pieces1, cv.COLOR_BGR2GRAY)

# Apply threshold to separate the pieces from the background
_, thresh = cv.threshold(gray, 254, 255, cv.THRESH_BINARY_INV) 

# Find the contours of the pieces
contours, _ = cv.findContours(thresh, cv.RETR_EXTERNAL, cv.CHAIN_APPROX_SIMPLE)

# Calculate the area of all contours
areas = [cv.contourArea(c) for c in contours]

# Calculate the average area excluding very small pieces (to avoid outliers)
mean_area = np.mean([a for a in areas if a > 1000])  # Only consider areas > 1000
min_area = mean_area * 0.5  # Set the threshold to 50% of the average area

print(f"Average piece area: {mean_area:.2f}, Automatic threshold: {min_area:.2f}")

# Filter contours based on the automatic threshold
pieces_1 = []
for contour in contours:
    area = cv.contourArea(contour)
    if area > min_area:  # If the area is large enough, consider it as a piece
        x, y, w, h = cv.boundingRect(contour)
        piece = pieces1[y:y+h, x:x+w]
        pieces_1.append(piece)

# Sort the pieces by position for a more organized layout
pieces_1 = sorted(pieces_1, key=lambda p: p.shape[1], reverse=True)

# Print the number of pieces found after filtering
print(f"Number of pieces found after filtering: {len(pieces_1)}")

# Display 
fig, axes = plt.subplots(2, 5, figsize=(12, 6))  # Grid 2x5
fig.suptitle("Extracted Puzzle Pieces", fontsize=16)

for i, ax in enumerate(axes.flat):
    if i < len(pieces_1):
        ax.imshow(cv.cvtColor(pieces_1[i], cv.COLOR_BGR2RGB))
        ax.set_title(f"Piece {i+1}")
    ax.axis("off")

plt.show()
```

# we have to remove the background

```python
def remove_white_background(piece):
    # Convert to HSV color space
    hsv = cv.cvtColor(piece, cv.COLOR_BGR2HSV)

    # Define the white color range in HSV
    lower_white = np.array([0, 0, 200], dtype=np.uint8)
    upper_white = np.array([180, 30, 255], dtype=np.uint8)

    # Create a mask 
    mask = cv.inRange(hsv, lower_white, upper_white)

    # Invert the mask
    piece_mask = cv.bitwise_not(mask)

    # Refine the mask using morphological operations
    kernel = np.ones((3, 3), np.uint8)
    refined_mask = cv.morphologyEx(piece_mask, cv.MORPH_CLOSE, kernel)
    refined_mask = cv.morphologyEx(refined_mask, cv.MORPH_OPEN, kernel)

    # Apply the mask to get only the pieces
    piece_no_bg = cv.bitwise_and(piece, piece, mask=refined_mask)

    # Add an alpha channel for transparency
    piece_rgba = cv.cvtColor(piece_no_bg, cv.COLOR_BGR2BGRA)
    piece_rgba[:, :, 3] = refined_mask 

    return piece_rgba, refined_mask
```

```python
# sum the segmented pieces
pieces = pieces_1 + pieces_2

transparent_pieces = []
masks = []

for piece in pieces:
    piece_rgba, piece_mask = remove_white_background(piece)
    transparent_pieces.append(piece_rgba)
    masks.append(piece_mask)
```

# MATCHING

```python
sift = cv.SIFT_create(nfeatures = 0, nOctaveLayers = 3, contrastThreshold = 0.04, edgeThreshold = 10, sigma = 1.6)
gray_image = cv.cvtColor(peppa, cv.COLOR_BGR2GRAY)

# Find the keypoints and descriptors for the image
img_keypoints, img_descriptors = sift.detectAndCompute(gray_image, None)

# Draw keypoints on the image
image_with_keypoints = cv.drawKeypoints(peppa, img_keypoints, None)
plt.imshow(cv.cvtColor(image_with_keypoints, cv.COLOR_BGR2RGB))
plt.title("Image with SIFT Keypoints")
plt.axis('off')
plt.show()

print(f"Number of keypoints detected: {len(img_keypoints)}")
print(f"Descriptor size: {img_descriptors.shape}")

numero_di_pezzi = len(pieces_1 + pieces_2)

print(f"Number of elements in the list: {numero_di_pezzi}")
```

```python
sift = cv.SIFT_create(nfeatures = 0, nOctaveLayers = 3, contrastThreshold = 0.04, edgeThreshold = 10, sigma = 1.6)
gray_image = cv.cvtColor(peppa, cv.COLOR_BGR2GRAY)
 

pieces_keypoints = []
pieces_descriptors = []

for i, piece in enumerate(pieces): 
    keypoints, descriptors = sift.detectAndCompute(piece, None)
    pieces_keypoints.append(keypoints)
    pieces_descriptors.append(descriptors)
    
    # show the keypoints
    piece_with_keypoints = cv.drawKeypoints(piece, keypoints, None)
    plt.figure(figsize=(5, 5))
    plt.imshow(cv.cvtColor(piece_with_keypoints, cv.COLOR_BGR2RGB))
    plt.title(f"Pezzo {i + 1} con SIFT Keypoints")
    plt.axis('off')
    plt.show()


if img_descriptors is not None:
    print(f"Numero di keypoints rilevati: {len(img_keypoints)}")
    print(f"Dimensione dei descrittori: {img_descriptors.shape}")
else:
    print("Errore: Nessun descrittore trovato nell'immagine principale.")
```

```python
# Find the matches between the descriptors of the image and those of the pieces
ratio = 3  # Configurable Ratio Threshold for refining
matches = []

for i, p_descriptors in enumerate(pieces_descriptors):
    if p_descriptors is None or img_descriptors is None:
        print(f"Skipping piece {i+1} - No descriptors found.")
        continue
    
    # Find the matches using BFMatcher
    bf = cv.BFMatcher(cv.NORM_L2, crossCheck=True)
    matches_piece = bf.match(p_descriptors, img_descriptors)

    # Filter out matches with distance 0 (potentially incorrect)
    filtered_matches = [match for match in matches_piece if match.distance > 0]

    # Refine the matches using the ratio threshold
    if filtered_matches:
        min_distance = min([match.distance for match in filtered_matches])
        refined_matches = [match for match in matches_piece if match.distance < ratio * min_distance]
        matches.append(refined_matches)

        # Display
        matches_image = cv.drawMatches(pieces[i], pieces_keypoints[i], peppa, img_keypoints, refined_matches, None)
        plt.figure(figsize=(10, 10))
        plt.imshow(cv.cvtColor(matches_image, cv.COLOR_BGR2RGB))
        plt.title(f"Refined Matches for Piece {i + 1}")
        plt.axis('off')
        plt.show()
    else:
        matches.append([])
        print(f"No valid matches for piece {i+1}")
```

```python
# I calculate the Affine Transformation using the refined matches
affine_transforms = []
canvas = np.zeros((peppa.shape[0], peppa.shape[1], 4), dtype=np.uint8)

for i, refined_matches in enumerate(matches):
    if len(refined_matches) >= 3:  # At least 3 matches are needed for an affine transformation
        piece_points = np.float32([pieces_keypoints[i][match.queryIdx].pt for match in refined_matches]).reshape(-1, 1, 2)
        image_points = np.float32([img_keypoints[match.trainIdx].pt for match in refined_matches]).reshape(-1, 1, 2)

        # Calculate the affine transformation using the refined matches
        transform, inliers = cv.estimateAffinePartial2D(piece_points, image_points, method=cv.RANSAC, ransacReprojThreshold=3.0)

        if transform is not None:
            affine_transforms.append(transform)

            # Apply the affine transformation to the piece and the mask
            transformed_piece = cv.warpAffine(transparent_pieces[i], transform, (peppa.shape[1], peppa.shape[0]))
            transformed_mask = cv.warpAffine(masks[i], transform, (peppa.shape[1], peppa.shape[0]))

            # Add alpha channel (transparency)
            alpha_channel = transformed_mask
            transformed_piece_rgba = transformed_piece.copy()
            transformed_piece_rgba[:, :, 3] = alpha_channel

            # Combine the pieces on the canvas using alpha blending
            canvas[transformed_mask > 0] = transformed_piece_rgba[transformed_mask > 0]

            # Display 
            print(f"Transformation matrix for piece {i + 1}:")
            print(transform)
            plt.figure(figsize=(6, 6))
            plt.imshow(cv.cvtColor(transformed_piece_rgba, cv.COLOR_BGRA2RGBA))
            plt.title(f"Piece {i+1} Transformed with Refined Matches")
            plt.axis("off")
            plt.show()
        else:
            print(f"No valid transformation found for piece {i + 1}")
    else:
        print(f"Not enough matches for piece {i + 1} to compute affine transformation.")
```

```python
# Mostra il puzzle finale ricostruito
plt.figure(figsize=(10, 6))
plt.imshow(cv.cvtColor(canvas, cv.COLOR_BGRA2RGBA))
plt.title("Puzzle Ricostruito con Affine Transformation Raffinata")
plt.axis("off")
plt.show()
```
